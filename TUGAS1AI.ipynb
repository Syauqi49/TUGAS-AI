# **1. STUDI KASUS ELECTRICAL RESISTIVITY TOMOGRAPHY (ERT)**

### **Deskripsi**

Pada studi kasus ini, hasil **Electrical Resistivity Tomography (ERT)**
digunakan untuk memahami kondisi bawah permukaan pada daerah yang
berpotensi mengalami longsor.

Analisis diarahkan untuk mengenali perubahan resistivitas dan
memperkirakan keberadaan bidang gelincir berdasarkan kontras
sifat kelistrikan material.

### **Role**

> Anda berperan sebagai pendamping geofisika yang membantu
> menginterpretasikan hasil ERT untuk memahami kondisi bawah
> permukaan pada daerah rawan longsor.

### **Task**

> Analisis penampang resistivitas 2D untuk mengidentifikasi zona
> yang memiliki perubahan resistivitas dan memperkirakan kemungkinan
> posisi bidang gelincir.

### **Context**

> Bidang gelincir merupakan salah satu bagian penting dalam
> analisis kestabilan lereng. Perbedaan material, kandungan air,
> dan kondisi bawah permukaan dapat menghasilkan kontras resistivitas
> yang terlihat pada hasil ERT.

### **Input**

> Data yang diberikan berupa penampang resistivitas 2D, nilai
> resistivitas, kedalaman, posisi elektroda, konfigurasi pengukuran,
> serta informasi geologi atau kondisi longsor apabila tersedia.

### **Constraints**

> Jangan membuat nilai resistivitas atau kedalaman yang tidak terdapat
> pada data. Jangan menentukan bidang gelincir hanya berdasarkan satu
> nilai resistivitas. Interpretasi harus mempertimbangkan pola
> resistivitas dan informasi pendukung.

### **Output**

> Jelaskan pola distribusi resistivitas, zona dengan kontras yang
> menonjol, kemungkinan perubahan material, dan perkiraan posisi
> bidang gelincir apabila dapat diidentifikasi.

### **Quality Criteria**

> Interpretasi harus sesuai dengan pola pada penampang, menggunakan
> istilah geofisika yang tepat, dan memisahkan hasil pengamatan
> dari interpretasi.

### **Verification**

> Bandingkan hasil interpretasi dengan kondisi geologi, topografi,
> informasi lapangan, atau data pendukung lainnya untuk memeriksa
> kesesuaian perkiraan bidang gelincir.

# STUDI KASUS 2
# IP - MINERALISASI
# **2. STUDI KASUS INDUCED POLARIZATION (IP)**

### **Deskripsi**

Pada studi kasus ini, data **Induced Polarization (IP)** digunakan
untuk membantu mengenali zona bawah permukaan yang memiliki respons
polarisasi berbeda dan berpotensi berkaitan dengan mineralisasi.

### **Role**

> Anda berperan sebagai pendamping geofisika yang membantu
> menginterpretasikan respons IP dalam mengidentifikasi zona
> yang memiliki karakteristik berbeda dari lingkungan sekitarnya.

### **Task**

> Analisis distribusi nilai chargeability dan resistivitas untuk
> menemukan zona anomali yang dapat menjadi indikasi adanya
> perubahan material atau mineralisasi.

### **Context**

> Metode IP mengukur respons polarisasi material bawah permukaan.
> Nilai chargeability dan resistivitas dapat memberikan informasi
> tambahan mengenai karakteristik material ketika diinterpretasikan
> secara bersama-sama.

### **Input**

> Data yang diberikan berupa penampang chargeability, resistivitas,
> kedalaman, posisi pengukuran, serta informasi geologi atau
> mineralisasi jika tersedia.

### **Constraints**

> Jangan menyatakan suatu zona pasti merupakan mineralisasi hanya
> berdasarkan nilai chargeability atau resistivitas. Interpretasi
> harus mempertimbangkan pola kedua parameter dan kondisi geologi.

### **Output**

> Identifikasi zona anomali chargeability dan resistivitas,
> jelaskan hubungan keduanya, dan tentukan zona yang perlu
> diperhatikan dalam interpretasi mineralisasi.

### **Quality Criteria**

> Interpretasi harus mempertimbangkan chargeability dan resistivitas
> secara bersama-sama serta menjelaskan alasan geofisika dari
> setiap zona yang diinterpretasikan.

### **Verification**

> Periksa kesesuaian zona anomali dengan informasi geologi,
> data alterasi, data geokimia, atau data pemboran apabila tersedia.

# STUDI KASUS 3

# **3. STUDI KASUS SEISMIK REFRAKSI**

### **Deskripsi**

Pada studi kasus ini, hasil **seismik refraksi** digunakan untuk
menggambarkan susunan lapisan bawah permukaan berdasarkan
perbedaan kecepatan gelombang seismik.

### **Role**

> Anda berperan sebagai pendamping geofisika yang membantu
> menginterpretasikan model kecepatan hasil survei seismik refraksi.

### **Task**

> Analisis model kecepatan dan kedalaman untuk menentukan
> pembagian lapisan bawah permukaan serta batas antar lapisan.

### **Context**

> Perbedaan sifat fisik material bawah permukaan dapat menyebabkan
> perbedaan kecepatan rambat gelombang seismik. Informasi tersebut
> dapat digunakan untuk memperkirakan susunan lapisan bawah permukaan.

### **Input**

> Data yang diberikan berupa kurva travel time, model kecepatan,
> kedalaman batas lapisan, posisi geophone, serta informasi geologi
> jika tersedia.

### **Constraints**

> Jangan membuat nilai kecepatan atau kedalaman baru. Jangan
> menentukan jenis batuan secara mutlak hanya berdasarkan nilai
> kecepatan karena satu rentang kecepatan dapat dimiliki oleh
> beberapa jenis material.

### **Output**

> Jelaskan jumlah lapisan yang dapat dibedakan, variasi kecepatan,
> kedalaman batas lapisan, dan kemungkinan karakter material
> berdasarkan informasi yang tersedia.

### **Quality Criteria**

> Interpretasi harus konsisten dengan model kecepatan dan kedalaman
> serta menjelaskan dasar yang digunakan untuk membedakan setiap lapisan.

### **Verification**

> Periksa kembali interpretasi terhadap kurva travel time, model
> kecepatan, dan informasi geologi pendukung.

# STUDI KASUS 4
# MAGNETOTELLURIK - PANAS BUMI

# **4. STUDI KASUS MAGNETOTELLURIK (MT)**

### **Deskripsi**

Pada studi kasus ini, metode **Magnetotellurik (MT)** digunakan
untuk memperoleh gambaran resistivitas bawah permukaan pada daerah
yang memiliki potensi sistem panas bumi.

### **Role**

> Anda berperan sebagai pendamping geofisika yang membantu
> menginterpretasikan model resistivitas hasil survei MT.

### **Task**

> Analisis distribusi resistivitas terhadap kedalaman untuk
> mengenali zona konduktif dan resistif yang dapat memberikan
> informasi mengenai sistem panas bumi.

### **Context**

> Sistem panas bumi memiliki kondisi bawah permukaan yang kompleks.
> Variasi resistivitas dapat berkaitan dengan perubahan litologi,
> fluida, alterasi, maupun struktur geologi sehingga interpretasinya
> perlu dilakukan secara terpadu.

### **Input**

> Data yang diberikan berupa model resistivitas 1D atau 2D,
> nilai resistivitas, kedalaman, posisi titik pengukuran, serta
> informasi geologi dan manifestasi panas bumi apabila tersedia.

### **Constraints**

> Jangan menentukan litologi hanya berdasarkan nilai resistivitas.
> Zona konduktif tidak boleh langsung dianggap sebagai reservoir
> atau zona alterasi tanpa dukungan informasi lainnya.

### **Output**

> Jelaskan distribusi resistivitas terhadap kedalaman, identifikasi
> zona konduktif dan resistif, serta kemungkinan hubungan zona
> tersebut dengan komponen sistem panas bumi.

### **Quality Criteria**

> Interpretasi harus mempertimbangkan kedalaman, pola spasial,
> kontras resistivitas, dan informasi geologi yang tersedia.

### **Verification**

> Bandingkan model resistivitas dengan data manifestasi panas bumi,
> geologi, struktur, geokimia, atau data sumur jika tersedia.

# **5. STUDI KASUS GRAVITASI**

### **Deskripsi**

Pada studi kasus ini, data gravitasi digunakan untuk memahami
variasi densitas bawah permukaan dan mengenali kemungkinan
struktur geologi berdasarkan pola anomali.

### **Role**

> Anda berperan sebagai pendamping geofisika yang membantu
> menginterpretasikan pola anomali gravitasi.

### **Task**

> Analisis pola anomali gravitasi untuk mengenali zona yang
> menunjukkan perubahan densitas dan kemungkinan keberadaan
> struktur bawah permukaan.

### **Context**

> Variasi densitas material bawah permukaan dapat menghasilkan
> perubahan medan gravitasi. Pola anomali gravitasi dapat digunakan
> sebagai informasi awal untuk memahami kondisi geologi bawah permukaan.

### **Input**

> Data yang diberikan berupa peta anomali Bouguer, nilai anomali,
> koordinat atau posisi pengukuran, serta peta geologi atau
> informasi struktur jika tersedia.

### **Constraints**

> Jangan menentukan jenis batuan hanya berdasarkan anomali positif
> atau negatif. Jangan membuat nilai anomali baru dan pertimbangkan
> kemungkinan adanya lebih dari satu sumber anomali.

### **Output**

> Jelaskan pola anomali tinggi dan rendah, kemungkinan perubahan
> densitas, serta indikasi struktur geologi yang dapat dikaitkan
> dengan pola anomali tersebut.

### **Quality Criteria**

> Interpretasi harus memperhatikan pola spasial anomali dan
> mempertimbangkan informasi geologi yang tersedia sebagai
> pendukung interpretasi.

### **Verification**

> Bandingkan hasil interpretasi dengan peta geologi, data
> geofisika lainnya, atau informasi struktur bawah permukaan
> apabila tersedia.
